# Bài 3 - Vòng lặp `for` và `while`

**Mục tiêu:** cho máy tính **lặp lại** một công việc nhiều lần. Đây là sức mạnh lớn nhất của máy tính:
làm một việc hàng triệu lần trong nháy mắt mà không bao giờ chán.

Trong sơ đồ khối, vòng lặp là **mũi tên quay ngược lên trên**.

````python
# Ô CHUẨN BỊ - em chỉ cần chạy ô này (Shift + Enter), chưa cần hiểu hết.
def kiem_tra(ham, cac_truong_hop):
    """Chạy hàm của em với nhiều đầu vào và so sánh với kết quả mong đợi."""
    dung = 0
    for dau_vao, mong_doi in cac_truong_hop:
        if not isinstance(dau_vao, tuple):
            dau_vao = (dau_vao,)
        tham_so = ", ".join(repr(x) for x in dau_vao)
        try:
            ket_qua = ham(*dau_vao)
        except Exception as loi:
            ket_qua = f"LỖI: {loi}"
        dat = ket_qua == mong_doi
        dung += dat
        print(("  ĐÚNG " if dat else "  SAI  ") + f"{ham.__name__}({tham_so}) = {ket_qua!r}   (mong đợi {mong_doi!r})")
    tong = len(cac_truong_hop)
    print(f"\n=> Em làm đúng {dung}/{tong} trường hợp.", "Tuyệt vời!" if dung == tong else "Thử sửa lại nhé, em làm được mà!")

print("Đã chuẩn bị xong công cụ kiểm tra bài tập.")
````

## 1. Vòng lặp `for` với `range`

`for i in range(5):` nghĩa là: *cho `i` lần lượt nhận các giá trị 0, 1, 2, 3, 4, mỗi lần làm các lệnh thụt lề bên dưới*.

| Cách viết | Các giá trị của `i` |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 (bắt đầu từ 0, **không** gồm 5) |
| `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(0, 11, 2)` | 0, 2, 4, 6, 8, 10 (bước nhảy 2) |
| `range(10, 0, -3)` | 10, 7, 4, 1 (đếm ngược) |

````python
for i in range(5):
    print("Lần thứ", i)

print("---")
for i in range(0, 11, 2):
    print(i, end=" ")   # end=" " để in trên cùng một dòng
````

## 2. Ví dụ kinh điển: tính tổng 1 + 2 + 3 + ... + n

Ý tưởng: dùng biến `tong` như một **chiếc hộp tiết kiệm**. Ban đầu hộp rỗng (`tong = 0`), mỗi vòng bỏ thêm `i` vào hộp.

#### Sơ đồ khối: Tính tổng 1 + 2 + ... + n

Mũi tên từ `i = i + 1` **quay ngược lên** hình thoi: đó là vòng lặp. Vòng lặp dừng khi `i ≤ n` trở thành **Sai**.

<img src="https://raw.githubusercontent.com/donamvn/thcs-lap-trinh-python/main/so-do/03_tong_1_n.png" width="460">

<small>Tải sơ đồ: [ảnh PNG](https://raw.githubusercontent.com/donamvn/thcs-lap-trinh-python/main/so-do/03_tong_1_n.png) · [bản vẽ Excalidraw](https://github.com/donamvn/thcs-lap-trinh-python/blob/main/so-do/03_tong_1_n.excalidraw) · [mã Mermaid](https://github.com/donamvn/thcs-lap-trinh-python/blob/main/so-do/03_tong_1_n.mmd)</small>

````python
n = 5
tong = 0
for i in range(1, n + 1):
    tong = tong + i
    print(f"i = {i}:  tong = {tong}")   # in bảng theo dõi từng bước
print("Tổng từ 1 đến", n, "là", tong)
````

Những dòng in ra ở trên gọi là **bảng theo dõi (trace table)**: một kỹ năng rất quan trọng để hiểu thuật toán.
Em hãy tự lập bảng này ra giấy với `n = 4` rồi so sánh.

> **Chuyện thú vị:** nhà toán học **Gauss** khi còn nhỏ đã tìm ra công thức `n × (n + 1) / 2`. Hãy kiểm tra xem máy tính và Gauss có ra cùng kết quả không:

````python
n = 100
tong = 0
for i in range(1, n + 1):
    tong += i          # viết tắt của tong = tong + i
print("Máy tính cộng từng số:", tong)
print("Công thức Gauss:      ", n * (n + 1) // 2)
````

## 3. Vòng lặp lồng nhau: bảng cửu chương

````python
for a in range(2, 6):          # vòng ngoài: bảng 2 đến bảng 5
    print(f"--- Bảng {a} ---")
    for b in range(1, 11):     # vòng trong: nhân với 1 đến 10
        print(f"{a} x {b:2} = {a * b}")
````

**Vẽ hình bằng vòng lặp:** trong Python, `"*" * 3` cho ra `"***"`. Em thử đổi `chieu_cao` xem:

````python
chieu_cao = 5
for i in range(1, chieu_cao + 1):
    print(" " * (chieu_cao - i) + "*" * (2 * i - 1))
````

## 4. Vòng lặp `while`: lặp khi **chưa biết trước** số lần

`while dieu_kien:` nghĩa là *còn đúng thì còn lặp*. Dùng khi ta không biết trước phải lặp bao nhiêu lần.

**Ví dụ:** đếm số chữ số của một số nguyên dương `n`. Mỗi lần chia `n` cho 10 (lấy phần nguyên) là **bỏ đi một chữ số cuối**.

## 5. `break` và `continue`
- `break`: **thoát ngay** khỏi vòng lặp.
- `continue`: **bỏ qua** phần còn lại của lần lặp này, sang lần tiếp theo.

````python
# Tìm số đầu tiên từ 1 trở đi chia hết cho cả 7 và 13
i = 1
while True:
    if i % 7 == 0 and i % 13 == 0:
        print("Tìm thấy:", i)
        break
    i += 1

# In các số từ 1 đến 15, bỏ qua số chia hết cho 3
for i in range(1, 16):
    if i % 3 == 0:
        continue
    print(i, end=" ")
````
