# Bài 4 - Cấu trúc dữ liệu: danh sách, ngăn xếp, hàng đợi, từ điển

**Cấu trúc dữ liệu** là **cách sắp xếp dữ liệu** trong máy tính để dùng cho tiện.
Giống như ở nhà: quần áo xếp trong tủ, sách xếp trên giá, bát đĩa xếp chồng lên nhau. Mỗi cách xếp hợp với một kiểu sử dụng.

> *Thuật toán + Cấu trúc dữ liệu = Chương trình* (nhà khoa học máy tính Niklaus Wirth).

````python
# Ô CHUẨN BỊ - em chỉ cần chạy ô này (Shift + Enter), chưa cần hiểu hết.
def kiem_tra(ham, cac_truong_hop):
    """Chạy hàm của em với nhiều đầu vào và so sánh với kết quả mong đợi."""
    dung = 0
    for dau_vao, mong_doi in cac_truong_hop:
        if not isinstance(dau_vao, tuple):
            dau_vao = (dau_vao,)
        tham_so = ", ".join(repr(x) for x in dau_vao)
        try:
            ket_qua = ham(*dau_vao)
        except Exception as loi:
            ket_qua = f"LỖI: {loi}"
        dat = ket_qua == mong_doi
        dung += dat
        print(("  ĐÚNG " if dat else "  SAI  ") + f"{ham.__name__}({tham_so}) = {ket_qua!r}   (mong đợi {mong_doi!r})")
    tong = len(cac_truong_hop)
    print(f"\n=> Em làm đúng {dung}/{tong} trường hợp.", "Tuyệt vời!" if dung == tong else "Thử sửa lại nhé, em làm được mà!")

print("Đã chuẩn bị xong công cụ kiểm tra bài tập.")
````

## 1. Danh sách (`list`): dãy các hộp xếp thẳng hàng

Nếu một biến là **một hộp**, thì danh sách là **một dãy hộp** được đánh số thứ tự, **bắt đầu từ 0**.

```
 chỉ số:    0      1      2      3      4
          +----+ +----+ +----+ +----+ +----+
  diem =  | 7  | | 9  | | 5  | | 8  | | 10 |
          +----+ +----+ +----+ +----+ +----+
 chỉ số âm: -5     -4     -3     -2     -1
```

````python
diem = [7, 9, 5, 8, 10]
print("Cả danh sách:", diem)
print("Phần tử đầu tiên diem[0] =", diem[0])
print("Phần tử cuối cùng diem[-1] =", diem[-1])
print("Số phần tử len(diem) =", len(diem))
print("Lát cắt diem[1:4] =", diem[1:4])   # từ chỉ số 1 đến trước 4
````

| Thao tác | Code | Ý nghĩa |
|---|---|---|
| Thêm vào cuối | `a.append(x)` | thêm `x` vào cuối |
| Chèn | `a.insert(i, x)` | chèn `x` vào vị trí `i` |
| Xoá theo giá trị | `a.remove(x)` | xoá `x` đầu tiên gặp |
| Lấy ra | `a.pop()` / `a.pop(i)` | lấy ra phần tử cuối / ở vị trí `i` |
| Sửa | `a[i] = x` | thay phần tử thứ `i` |
| Kiểm tra có không | `x in a` | `True` nếu `x` có trong `a` |

````python
lop = ["An", "Bình", "Chi"]
lop.append("Dũng")        # bạn mới chuyển đến
lop.insert(0, "Anh")      # chèn lên đầu
lop.remove("Bình")        # Bình chuyển trường
lop[1] = "An (lớp trưởng)"
print(lop)
print("Chi" in lop, "Bình" in lop)
````

## 3. Ngăn xếp (Stack): vào sau, ra trước

Hãy nghĩ đến **chồng đĩa**: đĩa đặt vào **sau cùng** nằm **trên cùng**, và lấy ra **đầu tiên**.
Quy tắc này gọi là **LIFO** (*Last In, First Out*).

```
        Đặt vào (push)          Lấy ra (pop)
               \                  ^
                v                /
            +-----------------+
            |  Đĩa 3 (trên)   |   <- vào sau, ra trước
            +-----------------+
            |  Đĩa 2          |
            +-----------------+
            |  Đĩa 1 (dưới)   |
            +-----------------+
```

Ứng dụng quen thuộc: nút **Hoàn tác (Undo, Ctrl+Z)**, nút **Quay lại** của trình duyệt.
Trong Python, dùng `list` làm ngăn xếp: `append()` để đặt vào, `pop()` để lấy ra.

````python
lich_su = []                   # ngăn xếp rỗng
for thao_tac in ["gõ chữ A", "gõ chữ B", "tô đậm", "xoá dòng"]:
    lich_su.append(thao_tac)
    print("Làm:", thao_tac, "->", lich_su)

print("\nBấm Ctrl+Z hai lần:")
for _ in range(2):
    print("Hoàn tác:", lich_su.pop(), "->", lich_su)
````

**Bài toán kinh điển với ngăn xếp: kiểm tra dấu ngoặc.** Một biểu thức như `(2 + 3) * (4 - (1 + 1))` có ngoặc hợp lệ không?

Ý tưởng: gặp `(` thì đặt vào ngăn xếp; gặp `)` thì lấy một `(` ra. Nếu cần lấy mà ngăn xếp rỗng, hoặc cuối cùng vẫn còn `(`, thì **sai**.

````python
def ngoac_hop_le(s):
    ngan_xep = []
    for c in s:
        if c == "(":
            ngan_xep.append(c)
        elif c == ")":
            if len(ngan_xep) == 0:
                return False
            ngan_xep.pop()
    return len(ngan_xep) == 0

for bt in ["(2+3)*(4-(1+1))", "((1+2)", "(1+2))(", "()()"]:
    print(bt, "->", "hợp lệ" if ngoac_hop_le(bt) else "KHÔNG hợp lệ")
````

## 4. Hàng đợi (Queue): vào trước, ra trước

Giống **xếp hàng mua đồ ở căng tin**: ai đến trước được phục vụ trước. Quy tắc **FIFO** (*First In, First Out*).

```
  Vào cuối hàng -->  [ Chi ] [ Bình ] [ An ]  --> Ra ở đầu hàng
                                        ^ đến trước, được phục vụ trước
```

Ứng dụng: hàng đợi in tài liệu, danh sách phát nhạc, xử lý tin nhắn. Trong Python dùng `deque` (đọc là "đéc"):
`append()` để vào cuối hàng, `popleft()` để lấy người ở đầu hàng.

````python
from collections import deque

hang = deque()
for ban in ["An", "Bình", "Chi", "Dũng"]:
    hang.append(ban)
    print(ban, "vào hàng ->", list(hang))

while hang:                       # còn người trong hàng
    ban = hang.popleft()
    print("Phục vụ", ban, "-> còn lại", list(hang))
````

## 5. Từ điển (`dict`): tra cứu theo tên

**Từ điển** lưu các cặp **khoá → giá trị**, giống **danh bạ điện thoại**: tra *tên* (khoá) để ra *số điện thoại* (giá trị).
Tra cứu bằng khoá **rất nhanh**, không cần đi dò từng phần tử như danh sách.

````python
danh_ba = {"An": "0912 345 678", "Bình": "0987 654 321"}
danh_ba["Chi"] = "0901 111 222"        # thêm mới
print("Số của Bình:", danh_ba["Bình"])
print("Có số của Dũng không?", "Dũng" in danh_ba)

for ten, so in danh_ba.items():
    print(f"{ten:5} : {so}")

# Ứng dụng: đếm số lần xuất hiện của mỗi chữ cái
dem = {}
for c in "banana":
    dem[c] = dem.get(c, 0) + 1
print(dem)
````

## 6. So sánh các cấu trúc dữ liệu

| Cấu trúc | Hình ảnh đời thường | Quy tắc | Dùng khi |
|---|---|---|---|
| Danh sách `list` | Dãy ghế đánh số | Truy cập theo chỉ số | Lưu dãy dữ liệu có thứ tự |
| Ngăn xếp (stack) | Chồng đĩa | Vào sau ra trước (LIFO) | Hoàn tác, kiểm tra ngoặc |
| Hàng đợi (queue) | Xếp hàng căng tin | Vào trước ra trước (FIFO) | Phục vụ theo lượt |
| Từ điển `dict` | Danh bạ điện thoại | Tra theo khoá | Tra cứu nhanh theo tên |
