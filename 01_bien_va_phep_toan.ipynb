# Bài 1 - Biến, kiểu dữ liệu và phép toán

**Mục tiêu:** hiểu biến là gì, biết các kiểu dữ liệu cơ bản, dùng máy tính như một chiếc máy tính bỏ túi siêu mạnh.

Trước tiên hãy chạy **ô chuẩn bị** ngay bên dưới (ô này dùng để tự chấm bài tập).

````python
# Ô CHUẨN BỊ - em chỉ cần chạy ô này (Shift + Enter), chưa cần hiểu hết.
def kiem_tra(ham, cac_truong_hop):
    """Chạy hàm của em với nhiều đầu vào và so sánh với kết quả mong đợi."""
    dung = 0
    for dau_vao, mong_doi in cac_truong_hop:
        if not isinstance(dau_vao, tuple):
            dau_vao = (dau_vao,)
        tham_so = ", ".join(repr(x) for x in dau_vao)
        try:
            ket_qua = ham(*dau_vao)
        except Exception as loi:
            ket_qua = f"LỖI: {loi}"
        dat = ket_qua == mong_doi
        dung += dat
        print(("  ĐÚNG " if dat else "  SAI  ") + f"{ham.__name__}({tham_so}) = {ket_qua!r}   (mong đợi {mong_doi!r})")
    tong = len(cac_truong_hop)
    print(f"\n=> Em làm đúng {dung}/{tong} trường hợp.", "Tuyệt vời!" if dung == tong else "Thử sửa lại nhé, em làm được mà!")

print("Đã chuẩn bị xong công cụ kiểm tra bài tập.")
````

## 1. Biến là gì?

Hãy tưởng tượng **biến** là một **chiếc hộp có dán nhãn tên**. Ta cất một giá trị vào hộp, lúc cần thì gọi tên hộp để lấy ra.

```
    ten              tuoi
  +--------+       +------+
  |  "An"  |       |  13  |
  +--------+       +------+
```

Dấu `=` trong lập trình **không** có nghĩa là "bằng nhau" như trong Toán, mà là **"cất vào"** (phép gán):
lấy giá trị bên **phải**, cất vào hộp bên **trái**.

````python
ten = "An"
tuoi = 13
print(ten)
print(tuoi)
print("Bạn", ten, "năm nay", tuoi, "tuổi")
````

Hộp có thể được **thay giá trị mới**. Đoán xem dòng cuối in ra số mấy?

````python
tuoi = 13
tuoi = tuoi + 1   # lấy giá trị cũ (13) cộng 1 rồi cất lại vào hộp tuoi
print(tuoi)
````

## 2. Các kiểu dữ liệu cơ bản

| Kiểu | Tên trong Python | Ví dụ | Dùng cho |
|---|---|---|---|
| Số nguyên | `int` | `13`, `-5`, `0` | tuổi, số học sinh |
| Số thực | `float` | `8.5`, `3.14` | điểm trung bình, chiều cao |
| Chuỗi chữ | `str` | `"An"`, `"Hà Nội"` | tên, địa chỉ |
| Đúng / Sai | `bool` | `True`, `False` | đã nộp bài chưa? |

Lệnh `type(...)` cho biết một giá trị thuộc kiểu gì:

````python
print(type(13))
print(type(8.5))
print(type("An"))
print(type(True))
````

## 3. Phép toán

| Phép toán | Ký hiệu | Ví dụ | Kết quả |
|---|---|---|---|
| Cộng | `+` | `7 + 2` | `9` |
| Trừ | `-` | `7 - 2` | `5` |
| Nhân | `*` | `7 * 2` | `14` |
| Chia | `/` | `7 / 2` | `3.5` |
| Chia lấy phần nguyên | `//` | `7 // 2` | `3` |
| Chia lấy phần dư | `%` | `7 % 2` | `1` |
| Lũy thừa | `**` | `7 ** 2` | `49` |

````python
print(7 + 2, 7 - 2, 7 * 2, 7 / 2)
print(7 // 2, 7 % 2, 7 ** 2)
````

**Ví dụ đời thường về `//` và `%`:** Có **17 cái kẹo** chia đều cho **5 bạn**. Mỗi bạn được mấy cái, còn thừa mấy cái?

````python
keo = 17
so_ban = 5
moi_ban = keo // so_ban   # phần nguyên
con_thua = keo % so_ban   # phần dư
print("Mỗi bạn được", moi_ban, "cái kẹo, còn thừa", con_thua, "cái.")
````

## 6. Hàm: đóng gói thuật toán

Để dùng lại một thuật toán nhiều lần, ta gói nó vào một **hàm** bằng từ khóa `def`.
Hàm nhận **đầu vào** (tham số) và trả lại **đầu ra** bằng `return`. Các bài tập bên dưới đều yêu cầu em viết hàm.

````python
def dien_tich_hcn(dai, rong):
    return dai * rong

print(dien_tich_hcn(8, 5))
print(dien_tich_hcn(10, 3))
````
