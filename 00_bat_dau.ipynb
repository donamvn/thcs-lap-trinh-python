## 1. Notebook là gì?

Notebook giống một **quyển vở thông minh** gồm nhiều **ô (cell)**:

- **Ô chữ** (như ô em đang đọc): giải thích bài học.
- **Ô code** (có nút tròn **Chạy** hình tam giác bên trái): chứa lệnh Python mà máy tính sẽ chạy.

**Cách chạy một ô code:** bấm chuột vào ô rồi nhấn **`Shift + Enter`**. Kết quả hiện ngay bên dưới.

> Mẹo: chạy các ô **theo thứ tự từ trên xuống**. Nếu bị rối, vào menu *Thời gian chạy → Khởi động lại phiên* (Runtime → Restart session) rồi chạy lại từ đầu.

Hãy chạy ô dưới đây:

```python
print("Xin chào! Mình đang học lập trình Python.")
```

Lệnh `print(...)` nghĩa là **in ra màn hình**. Hãy so sánh hai dòng sau: một dòng có dấu ngoặc kép, một dòng không.
Em **đoán trước** kết quả rồi mới chạy nhé!

```python
print(2 + 3)      # không có ngoặc kép: máy TÍNH phép cộng
print("2 + 3")    # có ngoặc kép: máy in ĐÚNG NGUYÊN chữ
```

Phần sau dấu `#` là **chú thích**: máy bỏ qua, chỉ để người đọc hiểu.

**Thử thách nhỏ:** sửa ô dưới cho máy in ra tên và lớp của em.

```python
print("Tên em là: ...")
print("Em học lớp: ...")
```

## 2. Thuật toán là gì?

**Thuật toán** là một **dãy các bước rõ ràng, theo thứ tự** để giải quyết một bài toán.
Em dùng thuật toán hằng ngày mà không để ý: nấu mì, đánh răng, đi từ nhà đến trường...

Mọi thuật toán đều có 3 phần:

1. **Đầu vào (Input):** những gì ta có lúc đầu.
2. **Xử lý:** các bước biến đổi.
3. **Đầu ra (Output):** kết quả ta muốn.

## 3. Sơ đồ khối

**Sơ đồ khối** là cách **vẽ** thuật toán bằng các hình quy ước, giúp nhìn một lần là hiểu. Các lập trình viên thường vẽ sơ đồ **trước khi** viết code.

| Hình | Ý nghĩa | Màu trong tài liệu |
|---|---|---|
| Hình bầu dục | Bắt đầu / Kết thúc | Xanh lá |
| Hình bình hành | Nhập dữ liệu / Xuất kết quả | Xanh dương |
| Hình chữ nhật | Xử lý, tính toán | Vàng |
| Hình thoi | Điều kiện, câu hỏi Đúng/Sai | Đỏ nhạt |
| Mũi tên | Hướng đi của thuật toán | |


#### Sơ đồ khối: Pha mì tôm

## 4. Cách tự học hiệu quả với tài liệu này

1. **Đọc**: đọc phần giải thích và sơ đồ khối.
2. **Đoán**: trước khi chạy code, đoán xem kết quả là gì.
3. **Chạy**: nhấn `Shift + Enter` và so sánh với dự đoán.
4. **Sửa**: đổi số, đổi chữ, thậm chí cố tình làm sai để xem máy báo lỗi thế nào. **Lỗi là bạn tốt**: nó chỉ cho em biết chỗ cần sửa.
5. **Làm bài tập**: mỗi bài có ô `kiem_tra(...)` tự chấm điểm cho em.
