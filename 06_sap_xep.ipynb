# Bài 6 - Thuật toán sắp xếp

**Bài toán:** cho danh sách số, sắp xếp lại theo thứ tự **tăng dần**.
Ví dụ: `[5, 2, 9, 1, 6]` thành `[1, 2, 5, 6, 9]`.

Ta học ba thuật toán cơ bản, mỗi thuật toán là một **ý tưởng** khác nhau:

| Thuật toán | Ý tưởng đời thường |
|---|---|
| Nổi bọt (Bubble sort) | Bong bóng to nổi dần lên mặt nước |
| Chọn (Selection sort) | Chọn bạn thấp nhất đứng đầu hàng, rồi tiếp tục với số còn lại |
| Chèn (Insertion sort) | Xếp bài: rút từng lá, chèn vào đúng chỗ trên tay |

> Xem hoạt hình rất đẹp của các thuật toán này tại <https://visualgo.net/vi/sorting>.

````python
# Ô CHUẨN BỊ - em chỉ cần chạy ô này (Shift + Enter), chưa cần hiểu hết.
def kiem_tra(ham, cac_truong_hop):
    """Chạy hàm của em với nhiều đầu vào và so sánh với kết quả mong đợi."""
    dung = 0
    for dau_vao, mong_doi in cac_truong_hop:
        if not isinstance(dau_vao, tuple):
            dau_vao = (dau_vao,)
        tham_so = ", ".join(repr(x) for x in dau_vao)
        try:
            ket_qua = ham(*dau_vao)
        except Exception as loi:
            ket_qua = f"LỖI: {loi}"
        dat = ket_qua == mong_doi
        dung += dat
        print(("  ĐÚNG " if dat else "  SAI  ") + f"{ham.__name__}({tham_so}) = {ket_qua!r}   (mong đợi {mong_doi!r})")
    tong = len(cac_truong_hop)
    print(f"\n=> Em làm đúng {dung}/{tong} trường hợp.", "Tuyệt vời!" if dung == tong else "Thử sửa lại nhé, em làm được mà!")

print("Đã chuẩn bị xong công cụ kiểm tra bài tập.")
````

## 1. Sắp xếp nổi bọt (Bubble sort)

**Ý tưởng:** đi từ đầu đến cuối, so sánh **hai phần tử cạnh nhau**, nếu cặp nào **sai thứ tự** thì **đổi chỗ**.
Sau mỗi lượt, số lớn nhất còn lại sẽ "nổi" dần về cuối danh sách. Lặp lại cho đến khi xong.

**Đổi chỗ hai phần tử** trong Python rất gọn: `a[j], a[j + 1] = a[j + 1], a[j]`.

#### Sơ đồ khối: Sắp xếp nổi bọt

Có **hai vòng lặp lồng nhau**: vòng ngoài `i` đếm số lượt, vòng trong `j` đi so sánh từng cặp cạnh nhau.

<img src="https://raw.githubusercontent.com/donamvn/thcs-lap-trinh-python/main/so-do/06_noi_bot.png" width="460">

<small>Tải sơ đồ: [ảnh PNG](https://raw.githubusercontent.com/donamvn/thcs-lap-trinh-python/main/so-do/06_noi_bot.png) · [bản vẽ Excalidraw](https://github.com/donamvn/thcs-lap-trinh-python/blob/main/so-do/06_noi_bot.excalidraw) · [mã Mermaid](https://github.com/donamvn/thcs-lap-trinh-python/blob/main/so-do/06_noi_bot.mmd)</small>

````python
def noi_bot(a):
    a = a.copy()             # làm việc trên bản sao, không sửa danh sách gốc
    n = len(a)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
        print(f"Sau lượt {i + 1}: {a}")
    return a

print("Kết quả:", noi_bot([5, 2, 9, 1, 6]))
````

## 2. Sắp xếp chọn (Selection sort)

**Ý tưởng:** tìm số **nhỏ nhất** trong cả danh sách, đổi nó lên vị trí đầu tiên. Rồi tìm số nhỏ nhất trong phần còn lại, đổi lên vị trí thứ hai...
Giống như cô giáo xếp hàng: *"Bạn nào thấp nhất lên đứng đầu!"*

````python
def sap_xep_chon(a):
    a = a.copy()
    n = len(a)
    for i in range(n - 1):
        vt_min = i                       # giả sử a[i] nhỏ nhất
        for j in range(i + 1, n):
            if a[j] < a[vt_min]:
                vt_min = j
        a[i], a[vt_min] = a[vt_min], a[i]
        print(f"Lượt {i + 1}: đưa {a[i]} về vị trí {i} -> {a}")
    return a

print("Kết quả:", sap_xep_chon([5, 2, 9, 1, 6]))
````

## 3. Sắp xếp chèn (Insertion sort)

**Ý tưởng:** giống khi em **xếp bài trên tay**. Lần lượt rút từng lá, dịch các lá lớn hơn sang phải và **chèn** lá mới vào đúng chỗ.
Phần bên trái luôn là phần **đã sắp xếp**.

````python
def sap_xep_chen(a):
    a = a.copy()
    for i in range(1, len(a)):
        la_bai = a[i]
        j = i - 1
        while j >= 0 and a[j] > la_bai:   # dịch các lá lớn hơn sang phải
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = la_bai                 # chèn vào chỗ trống
        print(f"Chèn {la_bai}: {a[:i + 1]} | {a[i + 1:]}")
    return a

print("Kết quả:", sap_xep_chen([5, 2, 9, 1, 6]))
````

## 4. So sánh: thuật toán nào làm ít việc hơn?

Ta đếm số **phép so sánh** mỗi thuật toán phải làm với cùng một danh sách ngẫu nhiên.

````python
import random

def dem_noi_bot(a):
    a, dem = a.copy(), 0
    for i in range(len(a) - 1):
        for j in range(len(a) - 1 - i):
            dem += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return dem

def dem_chen(a):
    a, dem = a.copy(), 0
    for i in range(1, len(a)):
        x, j = a[i], i - 1
        while j >= 0:
            dem += 1
            if a[j] <= x:
                break
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = x
    return dem

random.seed(1)
print(f"{'n':>6} | {'Nổi bọt':>10} | {'Chèn':>10}")
for n in [10, 100, 1000]:
    a = [random.randint(1, 1000) for _ in range(n)]
    print(f"{n:>6} | {dem_noi_bot(a):>10,} | {dem_chen(a):>10,}")

gan_xong = list(range(1000)); gan_xong[10], gan_xong[11] = gan_xong[11], gan_xong[10]
print("\nDanh sách GẦN như đã sắp xếp (1000 phần tử):")
print("  Nổi bọt:", f"{dem_noi_bot(gan_xong):,}", " | Chèn:", f"{dem_chen(gan_xong):,}")
````

**Nhận xét:**
- Cả ba thuật toán đều làm khoảng `n × n / 2` phép so sánh khi dữ liệu lộn xộn: tăng **rất nhanh** khi `n` lớn.
- Sắp xếp chèn **cực nhanh** khi danh sách *gần như* đã sắp xếp.
- Python có sẵn `sorted()` dùng thuật toán **Timsort** thông minh hơn nhiều. Khi làm bài thực tế cứ dùng nó, nhưng hiểu cách tự sắp xếp giúp em **tư duy thuật toán**.

````python
diem = [7.5, 9.0, 5.5, 8.25, 10, 6.0]
print(sorted(diem))                  # tăng dần
print(sorted(diem, reverse=True))    # giảm dần

hoc_sinh = [("An", 8.5), ("Bình", 9.75), ("Chi", 7.0), ("Dũng", 9.75)]
xep_hang = sorted(hoc_sinh, key=lambda hs: hs[1], reverse=True)   # sắp theo điểm
for hang, (ten, d) in enumerate(xep_hang, start=1):
    print(f"Hạng {hang}: {ten} - {d} điểm")
````
