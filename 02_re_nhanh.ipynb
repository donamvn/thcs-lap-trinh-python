# Bài 2 - Câu lệnh rẽ nhánh `if`

**Mục tiêu:** giúp chương trình **ra quyết định**: nếu điều kiện đúng thì làm việc này, nếu sai thì làm việc khác.
Trong sơ đồ khối, rẽ nhánh chính là **hình thoi**.

Nhớ chạy ô chuẩn bị bên dưới trước nhé.

````python
# Ô CHUẨN BỊ - em chỉ cần chạy ô này (Shift + Enter), chưa cần hiểu hết.
def kiem_tra(ham, cac_truong_hop):
    """Chạy hàm của em với nhiều đầu vào và so sánh với kết quả mong đợi."""
    dung = 0
    for dau_vao, mong_doi in cac_truong_hop:
        if not isinstance(dau_vao, tuple):
            dau_vao = (dau_vao,)
        tham_so = ", ".join(repr(x) for x in dau_vao)
        try:
            ket_qua = ham(*dau_vao)
        except Exception as loi:
            ket_qua = f"LỖI: {loi}"
        dat = ket_qua == mong_doi
        dung += dat
        print(("  ĐÚNG " if dat else "  SAI  ") + f"{ham.__name__}({tham_so}) = {ket_qua!r}   (mong đợi {mong_doi!r})")
    tong = len(cac_truong_hop)
    print(f"\n=> Em làm đúng {dung}/{tong} trường hợp.", "Tuyệt vời!" if dung == tong else "Thử sửa lại nhé, em làm được mà!")

print("Đã chuẩn bị xong công cụ kiểm tra bài tập.")
````

## 1. Điều kiện: câu hỏi Đúng hoặc Sai

| Phép so sánh | Ý nghĩa | Ví dụ | Kết quả |
|---|---|---|---|
| `==` | bằng (hai dấu bằng!) | `5 == 5` | `True` |
| `!=` | khác | `5 != 3` | `True` |
| `>` `<` | lớn hơn, nhỏ hơn | `5 > 3` | `True` |
| `>=` `<=` | lớn hơn hoặc bằng, nhỏ hơn hoặc bằng | `5 <= 3` | `False` |

Ghép nhiều điều kiện: `and` (và, cả hai cùng đúng), `or` (hoặc, chỉ cần một cái đúng), `not` (phủ định).

````python
diem = 8.5
print(diem >= 8)
print(diem == 10)
print(diem >= 5 and diem < 8)
print(not (diem < 5))
````

## 3. `if` - `elif` - `else`: xếp loại học tập

Khi có **nhiều hơn 2 lựa chọn**, ta dùng `elif` (viết tắt của *else if*: "nếu không thì nếu"). Máy kiểm tra **lần lượt từ trên xuống**, gặp điều kiện đúng đầu tiên thì làm và **bỏ qua** các nhánh còn lại.

#### Sơ đồ khối: Xếp loại theo điểm trung bình



<img src="https://raw.githubusercontent.com/donamvn/thcs-lap-trinh-python/main/so-do/02_xep_loai.png" width="460">

````python
def xep_loai(d):
    if d >= 8:
        return "Tốt"
    elif d >= 6.5:
        return "Khá"
    elif d >= 5:
        return "Đạt"
    else:
        return "Chưa đạt"

for diem in [9.2, 7.0, 6.5, 5.0, 3.8]:
    print(diem, "->", xep_loai(diem))
````

````python
def giai_pt_bac_nhat(a, b):
    if a == 0:
        if b == 0:
            return "Vô số nghiệm"
        else:
            return "Vô nghiệm"
    else:
        return f"x = {-b / a}"

print(giai_pt_bac_nhat(2, -6))   # 2x - 6 = 0
print(giai_pt_bac_nhat(0, 5))    # 0x + 5 = 0
print(giai_pt_bac_nhat(0, 0))    # 0x + 0 = 0
````
